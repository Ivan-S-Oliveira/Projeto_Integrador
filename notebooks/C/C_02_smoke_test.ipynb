## Versão mínima do scikit-learn

A fábrica exige scikit-learn ≥ 1.8: a API de `l1_ratio` e o
`FrozenEstimator` (usado na calibração) só existem a partir dessa versão.
Falhar aqui evita que o ambiente só quebre lá no C_15.

```python
from importlib import metadata
from packaging.version import Version

SKLEARN_MIN = "1.8"

try:
    sklearn_version = metadata.version("scikit-learn")
except metadata.PackageNotFoundError:
    raise RuntimeError(
        f"scikit-learn>={SKLEARN_MIN} não está instalado. "
        f"Instale: pip install \"scikit-learn>={SKLEARN_MIN}\""
    ) from None

if Version(sklearn_version) < Version(SKLEARN_MIN):
    raise RuntimeError(
        f"scikit-learn>={SKLEARN_MIN} é necessário "
        f"(instalado: {sklearn_version}). "
        f"Atualize: pip install \"scikit-learn>={SKLEARN_MIN}\""
    )

print(f"scikit-learn: {sklearn_version} (>= {SKLEARN_MIN}) OK")
```
